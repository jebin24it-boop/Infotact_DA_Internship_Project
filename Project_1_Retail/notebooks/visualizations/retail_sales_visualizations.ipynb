# Retail Sales Visualization Dashboard 

This notebook focuses on building professional business visualizations to analyze retail sales trends, transaction activity, and customer purchasing behavior.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization Styling
sns.set_style("whitegrid")

# Load Dataset
df = pd.read_csv("../../data/final_cleaned_retail_dataset.csv")

df.head()
```

## Monthly Revenue Trend

Visualizing monthly revenue patterns to identify seasonal retail performance trends.

```python
monthly_revenue = df.groupby('Month')['TotalPrice'].sum()

plt.figure(figsize=(12,6))

sns.lineplot(
    x=monthly_revenue.index,
    y=monthly_revenue.values,
    marker='o',
    linewidth=3
)

plt.title("Monthly Revenue Trend", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Revenue", fontsize=12)

plt.xticks(rotation=0)

plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()

plt.show()
```

## Monthly Transaction Volume

Analyzing transaction counts across different months.

```python
monthly_transactions = df.groupby('Month')['InvoiceNo'].nunique()

plt.figure(figsize=(12,6))

sns.barplot(
    x=monthly_transactions.index,
    y=monthly_transactions.values
)

plt.title("Monthly Transaction Volume", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Transactions", fontsize=12)

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

## Weekday Revenue Analysis

Visualizing revenue contribution across weekdays to identify peak business periods.

```python
weekday_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Sunday'
]

weekday_revenue = df.groupby('DayName')['TotalPrice'].sum().reindex(weekday_order)

plt.figure(figsize=(12,6))

sns.barplot(
    x=weekday_revenue.index,
    y=weekday_revenue.values
)

plt.title("Weekday Revenue Analysis", fontsize=16, fontweight='bold')
plt.xlabel("Day", fontsize=12)
plt.ylabel("Revenue", fontsize=12)

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

## Hourly Customer Activity

Analyzing transaction activity across different hours of the day.

```python
hourly_transactions = df['Hour'].value_counts().sort_index()

plt.figure(figsize=(12,6))

sns.lineplot(
    x=hourly_transactions.index,
    y=hourly_transactions.values,
    marker='o',
    linewidth=3
)

plt.title("Hourly Customer Activity", fontsize=16, fontweight='bold')
plt.xlabel("Hour of Day", fontsize=12)
plt.ylabel("Transactions", fontsize=12)

plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()

plt.show()
```

## Revenue Distribution Analysis

Analyzing the distribution of transaction revenue across the retail dataset.

```python
plt.figure(figsize=(12,6))

sns.histplot(
    df['TotalPrice'],
    bins=40,
    kde=True
)

plt.title("Revenue Distribution", fontsize=16, fontweight='bold')
plt.xlabel("Total Price", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.xlim(0, 200)

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

## Top Countries by Revenue

Visualizing the highest revenue-generating countries within the retail dataset.

```python
top_countries = df.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title("Top 10 Countries by Revenue", fontsize=16, fontweight='bold')
plt.xlabel("Revenue", fontsize=12)
plt.ylabel("Country", fontsize=12)

plt.tight_layout()

plt.show()
```

## Top Selling Products Analysis

Visualizing the highest-performing products based on quantity sold and revenue generated.

```python
top_products_qty = (
    df.groupby('Description')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_products_qty.values,
    y=top_products_qty.index
)

plt.title("Top 10 Products by Quantity Sold", fontsize=16, fontweight='bold')
plt.xlabel("Quantity Sold")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
top_products_revenue = (
    df.groupby('Description')['TotalPrice']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_products_revenue.values,
    y=top_products_revenue.index
)

plt.title("Top 10 Products by Revenue", fontsize=16, fontweight='bold')
plt.xlabel("Revenue")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

## Country-wise Sales Performance

Analyzing sales contribution across different countries.

```python
country_revenue = (
    df.groupby('Country')['TotalPrice']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=country_revenue.values,
    y=country_revenue.index
)

plt.title("Top 10 Countries by Revenue", fontsize=16, fontweight='bold')
plt.xlabel("Revenue")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
country_transactions = (
    df.groupby('Country')['InvoiceNo']
    .nunique()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=country_transactions.values,
    y=country_transactions.index
)

plt.title("Top 10 Countries by Transactions", fontsize=16, fontweight='bold')
plt.xlabel("Transactions")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

# Customer Performance Analysis

This section identifies the highest-value customers based on revenue contribution and transaction frequency.

```python
top_customers = (
    df.groupby('CustomerID')['TotalPrice']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))

top_customers.plot(
    kind='bar',
    title='Top 10 Customers by Revenue'
)

plt.xlabel('Customer ID')
plt.ylabel('Revenue')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
top5 = top_customers.head(5)

plt.figure(figsize=(8,8))

plt.pie(
    top5.values,
    labels=top5.index,
    autopct='%1.1f%%'
)

centre_circle = plt.Circle((0,0),0.60,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title('Revenue Share of Top 5 Customers')

plt.show()
```
