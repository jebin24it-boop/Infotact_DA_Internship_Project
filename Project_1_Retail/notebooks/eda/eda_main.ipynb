## Country-wise Transaction Analysis

Analyzing transaction distribution and revenue contribution across different countries.

```python
country_transactions = df['Country'].value_counts().head(10)

print(country_transactions)
```

```python
country_revenue = df.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False).head(10)

print(country_revenue)
```

### Observation

The United Kingdom dominates both transaction volume and total revenue within the retail dataset, contributing significantly higher sales activity compared to all other countries. Germany, France, and EIRE also show notable customer engagement and revenue generation, but at a much smaller scale compared to the United Kingdom.

## Product Performance Analysis

Identifying products with the highest sales quantities and revenue generation.

```python
top_products_quantity = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)

print(top_products_quantity)
```

```python
top_products_revenue = df.groupby('Description')['TotalPrice'].sum().sort_values(ascending=False).head(10)

print(top_products_revenue)
```

### Observation

Products such as “PACK OF 72 RETROSPOT CAKE CASES” and “ASSORTED COLOUR BIRD ORNAMENT” recorded the highest sales quantities, indicating strong customer demand and frequent purchases. In terms of revenue generation, products like “PARTY BUNTING” and “WHITE HANGING HEART T-LIGHT HOLDER” contributed the highest overall sales value, highlighting their strong commercial performance within the retail dataset.

```python
print(df.groupby('Country')['InvoiceNo'].nunique().sort_values(ascending=False).head(10))
```

### Observation

The United Kingdom recorded the highest number of unique transactions, indicating significantly stronger customer purchasing activity compared to other countries. Germany, France, and EIRE also contributed notable transaction volumes, while the remaining countries showed comparatively lower retail activity.
