# Week 3 - Exploratory Data Analysis

## Dataset Loading

In this section, the cleaned retail dataset is loaded for exploratory data analysis and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../../data/final_cleaned_retail_dataset.csv")

df.head()
```

## Dataset Overview

This section explores the structure of the cleaned retail dataset, including dimensions, column names, and datatypes.

```python
print(df.shape)

print(df.columns)

df.info()
```

## Statistical Summary and Missing Values Analysis

Generating statistical summaries and verifying missing values after preprocessing and cleaning.

```python
df.describe()

df.isnull().sum()
```

## Country-wise Transaction Analysis

Analyzing transaction distribution and revenue contribution across different countries.

```python
country_transactions = df['Country'].value_counts().head(10)

print(country_transactions)
```

```python
country_revenue = df.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False).head(10)

print(country_revenue)
```

### Observation

The United Kingdom dominates both transaction volume and total revenue within the retail dataset, contributing significantly higher sales activity compared to all other countries. Germany, France, and EIRE also show notable customer engagement and revenue generation, but at a much smaller scale compared to the United Kingdom.

## Product Performance Analysis

Identifying products with the highest sales quantities and revenue generation.

```python
top_products_quantity = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)

print(top_products_quantity)
```

```python
top_products_revenue = df.groupby('Description')['TotalPrice'].sum().sort_values(ascending=False).head(10)

print(top_products_revenue)
```

### Observation

Products such as “PACK OF 72 RETROSPOT CAKE CASES” and “ASSORTED COLOUR BIRD ORNAMENT” recorded the highest sales quantities, indicating strong customer demand and frequent purchases. In terms of revenue generation, products like “PARTY BUNTING” and “WHITE HANGING HEART T-LIGHT HOLDER” contributed the highest overall sales value, highlighting their strong commercial performance within the retail dataset.

```python
print(df.groupby('Country')['InvoiceNo'].nunique().sort_values(ascending=False).head(10))
```

### Observation

The United Kingdom recorded the highest number of unique transactions, indicating significantly stronger customer purchasing activity compared to other countries. Germany, France, and EIRE also contributed notable transaction volumes, while the remaining countries showed comparatively lower retail activity.

## Monthly Revenue Trend Analysis

Analyzing monthly sales revenue and transaction trends to identify periods of higher customer purchasing activity.

```python
monthly_revenue = df.groupby('Month')['TotalPrice'].sum().sort_values(ascending=False)

print(monthly_revenue)

monthly_transactions = df.groupby('Month')['InvoiceNo'].nunique()

print(monthly_transactions)
```

## Weekday-wise Customer Activity Analysis

Analyzing customer purchasing behavior across different weekdays.

```python
day_transactions = df['DayName'].value_counts()

print(day_transactions)

day_revenue = df.groupby('DayName')['TotalPrice'].sum().sort_values(ascending=False)

print(day_revenue)
```

## Hour-wise Transaction Analysis

Analyzing transaction distribution across different hours to identify peak shopping periods.

```python
hourly_transactions = df['Hour'].value_counts().sort_index()

print(hourly_transactions)

hourly_revenue = df.groupby('Hour')['TotalPrice'].sum().sort_values(ascending=False)

print(hourly_revenue)
```

### Observation

The analysis shows that November generated the highest overall revenue and transaction volume, indicating a peak sales period within the retail dataset. Thursday recorded the highest customer activity and revenue contribution among all weekdays, suggesting increased purchasing behavior during that period.

Hourly transaction analysis reveals that customer activity and revenue generation peak during midday hours, particularly between 12 PM and 2 PM, highlighting the busiest shopping window within the dataset.

## Customer Purchase Analysis

Analyzing customer purchasing frequency and identifying top customers based on transaction activity and revenue contribution.

```python
top_customers_transactions = df.groupby('CustomerID')['InvoiceNo'].nunique().sort_values(ascending=False).head(10)

print(top_customers_transactions)
```

```python
top_customers_revenue = df.groupby('CustomerID')['TotalPrice'].sum().sort_values(ascending=False).head(10)

print(top_customers_revenue)
```

### Observation

The customer analysis reveals that a small group of customers contributes a significantly higher number of transactions and overall revenue compared to others. Customer ID 14911 generated the highest total revenue, while Customer ID 12748 recorded the highest transaction frequency, indicating strong repeat purchasing behavior among selected customers within the retail dataset.

## Quantity and Pricing Analysis

Analyzing product quantity distribution and unit price behavior across retail transactions.

```python
print(df['Quantity'].describe())
```

```python
print(df['UnitPrice'].describe())
```

```python
highest_priced_products = df[['Description', 'UnitPrice']].sort_values(by='UnitPrice', ascending=False).head(10)

print(highest_priced_products)
```

### Observation

The quantity analysis shows that most retail transactions involve relatively small purchase quantities, with an average quantity of approximately 7 units per transaction. The unit price analysis indicates that the majority of products are moderately priced, while a few premium products such as “PARISIENNE CURIO CABINET” and “WHITE WIRE EGG HOLDER” recorded the highest unit prices within the cleaned dataset.
